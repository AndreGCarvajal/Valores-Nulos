```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('BostonExt.csv')
data.head()
```

```python
#Identificar valores nulos por columna
valores_nulos=data.isnull().sum()
list(valores_nulos)

```

```python
for indice, columna in enumerate(data.columns):
  print(f'{indice}: {columna}')
```

```python
data2=data.copy()
```

```python
data2.info()

```

```python
valores_nulos = data2.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
#0. Borramos columnas inceserarias (las que tienen las columnas llenas de valores nulos "4378", los id inecesarios, las URL y datos replicados en texto)
# Opción tradicional con axis=1
data2.drop(['id','listing_url', 'scrape_id', 'description', 'neighborhood_overview', 'host_id', 'host_profile_id', 'host_since','host_response_time','host_response_rate','host_acceptance_rate','host_thumbnail_url','host_neighbourhood','host_total_listings_count','host_verifications','neighbourhood','instant_bookable','neighbourhood_group_cleansed','calendar_updated','bathrooms_text', 'picture_url','host_url','host_profile_url','host_picture_url'], axis=1, inplace=True)
valores_nulos = data2.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
# 1. Eliminar filas donde las variables del anfitrión tienen 18 nulos o menos
data2.dropna(subset=[
    'host_name', 
    'hosts_time_as_user_years', 
    'hosts_time_as_user_months', 
    'hosts_time_as_host_years', 
    'hosts_time_as_host_months', 
    'host_listings_count', 
    'host_has_profile_pic', 
    'host_identity_verified',
], inplace=True)

valores_nulos = data2.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
# 2. Imputar características numéricas con la mediana (redondeada a 1 decimal si es necesario)
data2["bathrooms"] = data2["bathrooms"].fillna(round(data2["bathrooms"].median(), 1))
data2["bedrooms"] = data2["bedrooms"].fillna(round(data2["bedrooms"].median(), 1))
data2["beds"] = data2["beds"].fillna(round(data2["beds"].median(), 1))

valores_nulos = data2.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
# 3. Imputar textos con "Sin Información"
columnas_texto = ['host_location', 'host_about', 'license']
data2[columnas_texto] = data2[columnas_texto].fillna('Sin información')

valores_nulos = data2.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
# 4. Eliminar filas sin precio
data2.dropna(subset=['price'], inplace=True)

valores_nulos = data2.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
# 5. Rellenar reseñas con 0 
columnas_reseñas = ['review_scores_rating', 'review_scores_accuracy', 'reviews_per_month', 'review_scores_cleanliness', 'review_scores_checkin', 'review_scores_communication', 'review_scores_location', 'review_scores_value', 'first_review', 'last_review']
data2[columnas_reseñas] = data2[columnas_reseñas].fillna(0)

valores_nulos = data2.isnull().sum()
valores_nulos[valores_nulos > 0]
```

```python
# Guardar el dataframe en un nuevo archivo CSV sin incluir el índice
data2.to_csv('Boston_Limpio.csv', index=False)
```
